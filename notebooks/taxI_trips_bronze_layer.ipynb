```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import current_timestamp
```

```python
spark = SparkSession.builder.appName("taxi_trip_data").getOrCreate()
df = spark.read.csv('/Volumes/taxi_trips/bronze/raw/*.csv', header=True, inferSchema=True)
```

```python
# Adicionado timestamp para identificar a hora de inserção da tabela
df = df.withColumn(
    "inserted_at",
    current_timestamp()
)
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("OverwriteSchema", "true") \
    .saveAsTable('taxi_trips.bronze.trips')
```
